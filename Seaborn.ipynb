```python
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
df = sb.load_dataset("tips")
df
```

```python
df["size"].unique()
```

## Histplot

```python
sb.histplot(df["total_bill"],bins = 20,kde = True)
```

```python
plt.subplot(2,1,1)
sb.histplot(df["total_bill"],bins = 20,kde = True)
plt.subplot(2,1,2)
sb.histplot(df["tip"],bins = 20,kde = True)
```

## Jointplot

```python
sb.jointplot(x= "total_bill", y = "tip", data = df,kind = "scatter")
# different types of kind
```

```python
sb.jointplot(x= "total_bill", y = "tip", data = df,kind = "reg")
# different types of kind
```

```python
sb.jointplot(x= "total_bill", y = "tip", data = df,kind = "kde")
```

```python
sb.jointplot(x= "total_bill", y = "tip", data = df,kind = "hist")
```

## pairplot

```python
# this function will create a grid of Axes such that each numeric
# variable in ``data`` will by shared across the y-axes across a single row and
# the x-axes across a single column

sb.pairplot(df)
```

```python
# IF WANT TO DIFFERENTIATE PROPERLY
sb.pairplot(df, hue = "sex")
```

##  rugplot

```python
sb.rugplot(df)
# AGAIN ONLY NUMERIC VALUES
```

# Categorical Plots

### Categorical Data Plots
Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

1) boxplot
2) violinplot
3) stripplot
4) swarmplot
5) barplot
6) countplot


## Countplot

```python
sb.countplot(x= df["sex"], hue = df["smoker"])
```

## Barplot

```python
sb.barplot(x = df["sex"] , y = df["tip"])
```

```python
import numpy as np
sb.barplot(y = df["total_bill"], x = df["sex"],estimator = np.sum)
```

## Boxplot

```python
sb.boxplot(x = "total_bill", y = "sex", data = df, palette = "rainbow")
# here , x = quantitative & y = categorical data
```

```python
sb.violinplot(y = "total_bill", x = "day", data = df, palette = "rainbow")
```

## Stripplot

```python
sb.stripplot(y = "total_bill", x = "day", data = df)
```

## Swarmplot

```python
sb.swarmplot(y = "total_bill", x = "day", data = df)
```

## Combining swarm and strip plots

```python
sb.swarmplot(y = "total_bill", x = "day", data = df)
sb.violinplot(y = "total_bill", x = "day", data = df)
```

```python
sb.violinplot(y = "total_bill", x = "day", data = df)
sb.swarmplot(y = "total_bill", x = "day", data = df)
```
