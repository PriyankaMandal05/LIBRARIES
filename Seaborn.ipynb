# Categorical Plots

### Categorical Data Plots
Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

1) boxplot
2) violinplot
3) stripplot
4) swarmplot
5) barplot
6) countplot


## Countplot

```python
sb.countplot(x= df["sex"], hue = df["smoker"])
```

## Barplot

```python
sb.barplot(x = df["sex"] , y = df["tip"])
```

```python
import numpy as np
sb.barplot(y = df["total_bill"], x = df["sex"],estimator = np.sum)
```

## Boxplot

```python
sb.boxplot(x = "total_bill", y = "sex", data = df, palette = "rainbow")
# here , x = quantitative & y = categorical data
```

```python
sb.violinplot(y = "total_bill", x = "day", data = df, palette = "rainbow")
```

## Stripplot

```python
sb.stripplot(y = "total_bill", x = "day", data = df)
```

## Swarmplot

```python
sb.swarmplot(y = "total_bill", x = "day", data = df)
```

## Combining swarm and strip plots

```python
sb.swarmplot(y = "total_bill", x = "day", data = df)
sb.violinplot(y = "total_bill", x = "day", data = df)
```

```python
sb.violinplot(y = "total_bill", x = "day", data = df)
sb.swarmplot(y = "total_bill", x = "day", data = df)
```

# Matrix Plots

Matrix plots inclused 3 main plots which are 
1)Heat map
2)Cluster map
3)Pivot table heat map

### Heat map

```python
tipscorrelation = df[["total_bill","tip","size"]]
```

```python
tipscorrelation
```

```python
tipscorrelation.corr()
```

```python
sb.heatmap(tipscorrelation.corr())
```

```python
sb.heatmap(tipscorrelation.corr(),annot = True)
```

## Cluster map

```python
sb.clustermap(tipscorrelation.corr())
```
