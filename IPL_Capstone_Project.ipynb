**Now, Here comes some Basic Questions**

1. Which team won the most matches?

```python
df.head()
```

```python
wins = df["match_winner"].value_counts()
wins
```

```python
sb.barplot(x = wins.values , y = wins.index)
## THEREFORE MOST MATCH WIN = GUJRAT
```

2. Toss Decision Trends

```python
df["toss_decision"].value_counts()
```

```python
sb.countplot(x = df["toss_decision"])
plt.title("Toss decision IPL")
```

3. Toss Winner vs Match Winner

```python
count = df[df["toss_winner"] == df["match_winner"]]["match_id"].count()
```

```python
count
```

```python
percentage = (count / df.shape[0])*100
## this is of above count variable
```

```python
percentage.round(3)
```

4. How do teams win? (Runs vs Wickets)

```python
sb.countplot(x = df["won_by"])
```

###  Key Player Performances

1 Most "Player of the Match" Awards

```python
award = df["player_of_the_match"].value_counts().head()
```

```python
sb.barplot(y = award.index , x = award.values)
plt.title("Man of the Match")
```

2 Top Scorers

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

10 Best Bowling Figures

```python
bowling_fig = df.groupby('best_bowling')['best_bowling_figure'].sum().sort_values(ascending=False).head(2)
bowling_fig
```
