# Matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### Basics First

```python
a = np.linspace(1,5,10)
a
b = a ** 2
b
```

```python
plt.plot(a,b) # HERE WE HAVE TO COMPULSORILY GIVE WHAT U DEFINED ONLY.
```

```python
plt.title("Numbers")
plt.xlabel("x-axis")
```

### subplot

```python
plt.subplot(4,4,1)
plt.plot(a,b)
plt.subplot(4,4,3)
plt.subplot(4,4,7)
plt.subplot(4,4,2)
plt.subplot(4,4,6)
# TOTAL 16 SUBPLOTS as 4 rows and 4 cols given
```

### object oriented way

```python
fig = plt.figure(figsize = (12,9),dpi = 100)
axis1 = fig.add_axes([0.5,0.1,1,0.8])
axis1.plot(a,b)
axis2 = fig.add_axes([0.1,0.1,0.5,0.5])
axis2.plot(a,b)
```

```python
plt.scatter(a,b)
```

```python
plt.boxplot(b)
```

### saving plots

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(a,b)
plt.savefig("New plot.png")
```
