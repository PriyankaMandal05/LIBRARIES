🟢 ROUND 4 — Boolean masking + where

This is extremely important for data science.

Take:

arr = np.array([10, 25, 40, 15, 60, 5])

Solve:

1.

Get all values greater than 20.

2.

Get all values between 10 and 40.

3.

Get all even numbers.

4.

Replace values greater than 30 with 0.

5.

Replace values less than 20 with -1.

6.

Find how many values are greater than 20.

You'll naturally use things like:

arr > 20
arr[arr > 20]

and eventually:

np.where(...)

That's much better than simply memorizing "np.where = condition".

🟢 ROUND 6 — Broadcasting

This deserves its own practice.

Start with:

Try:

arr + 10

Then:

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

Try:

arr + 10

Then:

arr + np.array([10, 20, 30])

And then deliberately create an incompatible shape and see the error.

Understanding why broadcasting works is much more valuable than memorizing the word.

🟢 ROUND 7 — Copy vs View

You already have this in your notes, so test it.

a = np.array([1, 2, 3])
b = a

Change b.

What happens to a?

Then:

a = np.array([1, 2, 3])
b = a.copy()

Change b.

What happens to a?

Then investigate slicing:

a = np.array([1, 2, 3, 4, 5])
b = a[1:4]

This is worth understanding because views vs copies can cause very confusing bugs later.
